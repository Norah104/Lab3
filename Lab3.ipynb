#Univariate Analysis

```python
plt.figure()
plt.hist(df['study_hours'], bins=20)
plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Frequency")
plt.show()
```

>The histogram shows how study hours are distributed among students.

#Bivariate Analysis

```python
plt.figure()
plt.scatter(df['study_hours'], df['productivity_score'])
plt.title("Study Hours vs Productivity Score")
plt.xlabel("Study Hours")
plt.ylabel("Productivity Score")
plt.show()
```

>The scatter plot shows the relationship between study hours and productivity score.
